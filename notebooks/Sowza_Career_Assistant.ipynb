# 📄 Etapa 5 — Carregamento da Base de Conhecimento

Nesta etapa são carregados todos os documentos da base de conhecimento utilizados pelo **Sowza Career Assistant**.

Os documentos estão armazenados na pasta **knowledge-base** do repositório e organizados no formato **Markdown (.md)**.

Cada arquivo é lido e armazenado em uma estrutura de dados contendo:

- Nome do arquivo de origem.
- Conteúdo completo do documento.

Essa estrutura servirá como entrada para a etapa de preparação dos documentos (Chunking).

## Objetivos desta etapa

- Localizar a pasta da base de conhecimento.
- Ler todos os arquivos Markdown.
- Armazenar os documentos em memória para processamento posterior.

````python
# ============================================================
# ETAPA 5 - Carregamento da Base de Conhecimento
# ============================================================

from pathlib import Path

# Caminho da base de conhecimento
knowledge_path = Path("Sowza-Career-Assistant-/knowledge-base")

documents = []

for file in sorted(knowledge_path.glob("*.md")):
    with open(file, "r", encoding="utf-8") as f:
        documents.append({
            "source": file.name,
            "content": f.read()
        })

print(f"✅ Total de documentos carregados: {len(documents)}")
````

# 🔍 Validação da Base de Conhecimento

Após o carregamento dos documentos, é importante verificar se todos os arquivos da base de conhecimento foram lidos corretamente.

Nesta etapa serão exibidos os nomes dos documentos encontrados, permitindo confirmar que a base está completa antes de iniciar o processamento dos textos.

````python
# ============================================================
# VALIDAÇÃO DA BASE DE CONHECIMENTO
# ============================================================

print("📄 Documentos carregados:\n")

for i, doc in enumerate(documents, start=1):
    print(f"{i:02d}. {doc['source']}")

print(f"\n✅ Total de documentos: {len(documents)}")
````

# ✂️ Etapa 6 — Preparação dos Documentos (Chunking)

Após o carregamento da base de conhecimento, os documentos precisam ser divididos em partes menores, chamadas de **chunks**.

Essa etapa é fundamental para o funcionamento de sistemas **Retrieval-Augmented Generation (RAG)**, pois permite que o mecanismo de busca recupere apenas os trechos mais relevantes para responder a uma pergunta, em vez de utilizar documentos inteiros.

O processo de divisão é realizado utilizando o **RecursiveCharacterTextSplitter**, que preserva o contexto entre os trechos por meio de uma pequena sobreposição (*chunk overlap*).

## Objetivos desta etapa

- Configurar o mecanismo de divisão dos documentos.
- Definir o tamanho dos chunks.
- Definir a sobreposição entre os chunks.
- Preparar os documentos para a geração dos embeddings.

````python
# ============================================================
# ETAPA 6 - Configuração do Text Splitter
# ============================================================

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

print("✅ Text Splitter configurado com sucesso!")
print(f"📄 Tamanho do chunk: {text_splitter._chunk_size} caracteres")
print(f"🔄 Sobreposição: {text_splitter._chunk_overlap} caracteres")
````

# 🧩 Geração dos Chunks

Com o **Text Splitter** configurado, cada documento da base de conhecimento será dividido em pequenos trechos (*chunks*).

Cada chunk armazenará:

- O conteúdo textual.
- O documento de origem.

Esses chunks serão utilizados posteriormente para gerar os embeddings e construir a base vetorial do projeto.

````python
# ============================================================
# ETAPA 6 - Criação dos Chunks
# ============================================================

chunks = []

for doc in documents:
    split_texts = text_splitter.split_text(doc["content"])

    for text in split_texts:
        chunks.append({
            "source": doc["source"],
            "content": text
        })

print(f"✅ Total de chunks criados: {len(chunks)}")
````

# 🔍 Validação dos Chunks

Após a divisão dos documentos, é importante verificar se os chunks foram gerados corretamente.

Nesta etapa serão realizadas três validações:

1. Visualizar o primeiro chunk gerado.
2. Identificar o documento de origem desse chunk.
3. Verificar a distribuição dos chunks entre os documentos da base de conhecimento.

Essas verificações garantem que o processo de preparação dos documentos foi executado corretamente antes da geração dos embeddings.

````python
# ============================================================
# ETAPA 6 - Validação dos Chunks
# ============================================================

from collections import Counter

print("=" * 70)
print("📄 PRIMEIRO CHUNK")
print("=" * 70)

print(chunks[0]["content"])

print("\n" + "=" * 70)
print("📁 DOCUMENTO DE ORIGEM")
print("=" * 70)

print(chunks[0]["source"])

print("\n" + "=" * 70)
print("📊 DISTRIBUIÇÃO DOS CHUNKS")
print("=" * 70)

distribution = Counter(chunk["source"] for chunk in chunks)

for source, total in sorted(distribution.items()):
    print(f"{source:<35} {total:>3} chunks")

print("\n" + "=" * 70)
print(f"✅ Total de chunks: {len(chunks)}")
````

# 🧠 Etapa 7 — Geração dos Embeddings

Após a divisão dos documentos em chunks, o próximo passo consiste em transformá-los em **embeddings**.

Embeddings são representações vetoriais que convertem o conteúdo textual em valores numéricos, permitindo que computadores comparem textos com base em seu significado semântico, e não apenas por palavras iguais.

No **Sowza Career Assistant**, será utilizado o modelo de embeddings do **Google Gemini**, responsável por transformar cada chunk em um vetor que será armazenado posteriormente na base vetorial.

Esses vetores constituem o núcleo do mecanismo de busca semântica utilizado pelo sistema RAG.

## Objetivos desta etapa

- Configurar o modelo de embeddings.
- Preparar a geração dos vetores.
- Validar o funcionamento do modelo antes da criação da base vetorial.

````python
# ============================================================
# ETAPA 7 - Configuração do Modelo de Embeddings
# ============================================================

embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)

print("✅ Modelo de Embeddings configurado com sucesso!")
print("🧠 Modelo utilizado: models/gemini-embedding-001")
````

````python
# ============================================================
# ETAPA 7 - Validação dos Embeddings
# ============================================================

texto_teste = "Olá, mundo!"

embedding = embeddings.embed_query(texto_teste)

print("✅ Embedding gerado com sucesso!")

print(f"\n📏 Dimensão do vetor: {len(embedding)}")

print("\n🔢 Primeiros 10 valores do vetor:")

for i, valor in enumerate(embedding[:10], start=1):
    print(f"{i:02d}: {valor:.6f}")
````

# 🗄️ Etapa 8 — Construção da Base Vetorial (ChromaDB)

Após a geração dos embeddings, os documentos precisam ser armazenados em uma **base vetorial**.

Neste projeto será utilizado o **ChromaDB**, um banco de dados otimizado para armazenar vetores e realizar buscas semânticas.

Cada chunk da base de conhecimento será indexado juntamente com seu embedding, permitindo recuperar rapidamente os trechos mais relevantes durante as consultas do sistema RAG.

## Objetivos desta etapa

- Criar a base vetorial.
- Indexar todos os chunks.
- Associar cada vetor ao documento de origem.
- Preparar a base para utilização pelo Retriever.

````python
# ============================================================
# ETAPA 8 - Criação da Base Vetorial (ChromaDB)
# ============================================================

persist_directory = "./chroma_db"

vectorstore = Chroma.from_texts(
    texts=[chunk["content"] for chunk in chunks],
    embedding=embeddings,
    metadatas=[
        {"source": chunk["source"]}
        for chunk in chunks
    ],
    persist_directory=persist_directory,
)

print("✅ Base vetorial criada com sucesso!")
print(f"📂 Diretório: {persist_directory}")
print(f"📄 Chunks indexados: {len(chunks)}")
````

# 🔍 Validação da Base Vetorial

Após a criação da base vetorial, é importante verificar se todos os documentos foram indexados corretamente.

Nesta etapa serão realizadas algumas validações para garantir que a base está pronta para utilização pelo Retriever.

## Validações realizadas

- Verificar se a base vetorial foi criada.
- Confirmar a quantidade de chunks indexados.
- Verificar o diretório de persistência.
- Confirmar que o objeto `vectorstore` está disponível para consultas.

````python
# ============================================================
# ETAPA 8 - Validação da Base Vetorial
# ============================================================

from pathlib import Path

print("=" * 70)
print("🗄️ BASE VETORIAL")
print("=" * 70)

print(f"Tipo do Vector Store: {type(vectorstore).__name__}")

print(f"Diretório: {persist_directory}")

print(f"Chunks indexados: {len(chunks)}")

print(f"Diretório existe? {Path(persist_directory).exists()}")

print("\n✅ Base vetorial validada com sucesso!")
````

# 🔎 Etapa 9 — Configuração do Retriever

Com a base vetorial criada, o próximo passo é configurar o **Retriever**.

O Retriever é responsável por realizar buscas semânticas na base vetorial, recuperando os trechos mais relevantes para responder a uma pergunta.

Diferentemente de uma busca tradicional por palavras-chave, o Retriever utiliza os embeddings para encontrar documentos com significado semelhante ao da consulta realizada.

Esses documentos recuperados serão utilizados posteriormente pelo modelo Gemini para gerar respostas contextualizadas.

## Objetivos desta etapa

- Configurar o mecanismo de recuperação semântica.
- Definir a estratégia de busca.
- Preparar o componente responsável por alimentar o modelo de linguagem com informações relevantes.

````python
# ============================================================
# ETAPA 9 - Configuração do Retriever
# ============================================================

retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)

print("✅ Retriever configurado com sucesso!")
print("🔎 Quantidade de documentos recuperados por consulta: 3")
````

# 🤖 Etapa 10 — Construção do Pipeline RAG

Com todos os componentes preparados, chegou o momento de construir o pipeline do **Sowza Career Assistant**.

O pipeline RAG (Retrieval-Augmented Generation) integra:

- Base de conhecimento
- Busca semântica (Retriever)
- Prompt personalizado
- Modelo Gemini

O funcionamento ocorre em quatro etapas:

1. O usuário envia uma pergunta.
2. O Retriever recupera os chunks mais relevantes.
3. O Prompt organiza essas informações.
4. O Gemini gera a resposta utilizando apenas o contexto recuperado.

## Objetivos desta etapa

- Criar o Prompt Template.
- Integrar Retriever e Gemini.
- Construir a cadeia RAG.

````python
# ============================================================
# ETAPA 10 - Prompt do Sowza Career Assistant
# ============================================================

prompt = ChatPromptTemplate.from_template("""
Você é o Sowza Career Assistant, assistente virtual da desenvolvedora Tatiane Souza.

Sua função é responder perguntas sobre sua trajetória profissional utilizando exclusivamente as informações fornecidas no contexto.

Regras:

- Não invente informações.
- Se a resposta não estiver no contexto, informe que ela não foi encontrada na base de conhecimento.
- Organize a resposta de forma clara e objetiva.
- Utilize listas quando necessário.
- Não mencione que recebeu um contexto interno.

Contexto:
{context}

Pergunta:
{question}

Resposta:
""")

print("✅ Prompt Template criado com sucesso!")
````

# 🤖 Configuração do Modelo de Linguagem (LLM)

Nesta etapa será configurado o modelo de linguagem responsável por gerar as respostas do **Sowza Career Assistant**.

Será utilizado o modelo **Gemini 3.6 Flash**, que receberá:

- a pergunta do usuário;
- o contexto recuperado pelo Retriever;
- as instruções definidas no Prompt Template.

O modelo utilizará essas informações para produzir respostas fundamentadas exclusivamente na base de conhecimento do projeto.

## Objetivos desta etapa

- Inicializar o modelo Gemini.
- Configurar o LLM que será utilizado pela cadeia RAG.
- Preparar a integração com o Prompt Template.

````python
# ============================================================
# ETAPA 10 - Configuração do Modelo Gemini
# ============================================================

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

print("✅ Modelo Gemini configurado com sucesso!")
print("🤖 Modelo: gemini-3.6-flash")
````

# 🔗 Construção da Cadeia RAG

Com todos os componentes configurados, será construída a cadeia RAG (Retrieval-Augmented Generation).

A cadeia integra:

1. **Retriever** → Recupera os documentos mais relevantes.
2. **Prompt Template** → Organiza o contexto e a pergunta.
3. **Modelo Gemini** → Gera a resposta.
4. **Output Parser** → Converte a resposta para texto simples.

A partir desta etapa, o **Sowza Career Assistant** estará pronto para responder perguntas utilizando exclusivamente a base de conhecimento construída durante o projeto.

````python
# ============================================================
# ETAPA 10 - Construção da Cadeia RAG
# ============================================================

from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser


def format_docs(docs):
    """
    Formata os documentos recuperados pelo Retriever.

    Cada trecho é acompanhado do nome do arquivo de origem,
    fornecendo mais contexto ao modelo Gemini.
    """

    return "\n\n".join(
        f"Fonte: {doc.metadata['source']}\n\n{doc.page_content}"
        for doc in docs
    )


rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough(),
    }
    | prompt
    | llm
    | StrOutputParser()
)

print("✅ Cadeia RAG criada com sucesso!")
````

````python
import time

def perguntar(pergunta):
    """
    Envia uma pergunta ao pipeline RAG,
    exibe documentos utilizados,
    tempo de resposta
    e trata erros da API.
    """

    print("=" * 80)
    print("❓ PERGUNTA")
    print("=" * 80)
    print(pergunta)

    docs = retriever.invoke(pergunta)

    fontes = sorted({doc.metadata["source"] for doc in docs})

    print("\n📚 Documentos consultados")
    print("-" * 80)

    for fonte in fontes:
        print(f"• {fonte}")

    inicio = time.time()

    try:

        resposta = rag_chain.invoke(pergunta)

        fim = time.time()

        print("\n💬 RESPOSTA")
        print("=" * 80)
        print(resposta)

        print(f"\n⏱ Tempo de resposta: {fim - inicio:.2f} segundos")

    except Exception as erro:

        print("\n❌ Erro ao consultar o modelo Gemini.")
        print("-" * 80)
        print(erro)

        print("\n💡 Possíveis causas:")
        print("• Limite diário da API atingido.")
        print("• Limite de requisições por minuto.")
        print("• Problema temporário na conexão.")
        print("• Chave da API inválida.")

    print("\n")
````
